## (a)

Als $X$ een $Bin(n,p)$-verdeling heeft, dan 
$$
\mathbb{P}(X=k) = 
{n \choose k} p^k (1-p)^{n-k}, 
\quad k=0,\dots,n.
$$
Er geldt dan dat $Y = n-X$ voldoet aan
$$
\mathbb{P}(Y=k) = 
\mathbb{P}(n-X=k) =
\mathbb{P}(X=n-k) = {n \choose k} p^{n-k} (1-p)^k, 
\quad k=0,\dots,n,
$$
en dus heeft $Y$ een $Bin(n,1-p)$-verdeling.

Het steekproefgemiddelde en de steekproefvariante van de steekproef $1,3,4,5,2,3,4,5,6,7$ worden in de volgende cell uitgerekend.

```python
import numpy as np
data = np.array([1,3,4,5,2,3,4,5,6,7])
m = sum(data)/len(data)
s2 = sum((data-m)**2)/(len(data)-1)
print(m,s2)
```

Het kan sneller via

```python
m = np.mean(data)
s2 = np.var(data, ddof = 1)
print(m, s2)
```
